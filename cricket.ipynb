```python
import numpy as np
import pandas as pd
```

```python
# import csv file
df = pd.read_csv("./cricketData.csv")
df
```

```python
# rename the columns to make more sense
```

```python
df.rename(columns={"Mat":"MatchesPlayed", "Inns":"Innings","NO":"NotOuts",
                   "HS":"HighestScore", "*Ave*":"AverageScore", "BF":"BallsFaced",
                   "SR":"StrikeRate", "100":"Centuries", "50":"HalfCenturies"}, inplace=True)
```

```python
df.head()
```

```python
df.isnull().any()
```

```python
df.loc[df["BallsFaced"]=="-", "BallsFaced"] = np.nan # there were values who had '-' for null
```

```python
df = df.replace(0, np.nan) # making the 0's in the dataframe as null
```

```python
# Balls faced column has a string value, thus put 0 in place of null values
```

```python
df["BallsFaced"] = df["BallsFaced"].fillna(0)
```

```python
df["BallsFaced"].isna().sum()
```

```python
# remove null values from "StrikeRate" columns
```

```python
df["StrikeRate"] = df["StrikeRate"].interpolate()
```

```python
df["StrikeRate"].isna().sum()
```

```python
df.isnull().any() # null values are handled successfully
```

```python
# remove duplicate values
```

```python
#search if duplicate values exists or not
df.duplicated().sum()
```

```python
# drop the duplicate values
df = df.drop_duplicates()
```

```python
df.duplicated().sum() # would return zero now!
```

```python
# Split up Span Date into Start-Date and End-Date
```

```python
df["Start-Date"] = df.loc[:, "Span"].str.split("-").str[0]
df["End-Date"] = df.loc[:, "Span"].str.split("-").str[1]
```

```python
df
```

```python
# add new column -> Career Length
```

```python
df["Career-Length"] = np.int64(df["End-Date"]) - np.int64(df["Start-Date"])
```

```python
df
```

```python
#Drop Span Columns
```

```python
df = df.drop(columns=['Span'])
```

```python
df
```

```python
# split player and the country code
```

```python
df["Country"] = df["Player"].str.split("(", regex=False).str[1]
df["Player"] = df["Player"].str.split("(").str[0]
```

```python
df["Country"] = df["Country"].str.replace(")", "", regex=False)
```

```python
# in case if you fucked up country data, here is the help!
dummy = pd.read_csv("cricketData.csv")
df["Player"] = dummy["Player"]
# df = df.drop(columns=["Country"])
```

```python
df
```

```python
# fix up data type of columns
```

```python
df['MatchesPlayed'].dtypes # this would likely give you a string
```

```python
# clean the MatchesPlayed Column. It contains * somewhere
df['MatchesPlayed'] = df['MatchesPlayed'].str.replace("*", '')
df['MatchesPlayed'] = np.int64(df['MatchesPlayed'])
df['MatchesPlayed'].dtypes
```

```python
df.info()
```

```python
# clean the HighestScore Column. It contains * somewhere
df['HighestScore'] = df['HighestScore'].str.replace("*", '')
df['HighestScore'] = np.int64(df['HighestScore'])
df['HighestScore'].dtypes
```

```python
# clean the AverageScore Column. It contains * somewhere
df['AverageScore'] = df['AverageScore'].str.replace("*", '')
df['AverageScore'] = np.float64(df['AverageScore'])
df['AverageScore'].dtypes
```

```python
# clean the BallsFaced Column. It contains * somewhere
df['BallsFaced'] = df['BallsFaced'].str.replace("+", '')
df['BallsFaced'] = np.float64(df['BallsFaced'])
df['BallsFaced'].dtypes
```

```python
df.loc[:, ['4s', '6s']] = df.loc[:, ['4s', '6s']].apply(lambda x: x.str.replace("+", ''))
```

```python
#convert into an integer
df['4s'] = np.int64(df['4s'])
df['6s'] = np.int64(df['6s'])
```

```python
df.info()
```

```python
df['Start-Date']  = pd.to_datetime(df['Start-Date'])
df['End-Date']  = pd.to_datetime(df['End-Date'])
```

```python
df.info()
```
